```python
from openpyxl import Workbook
```

```python
wb = Workbook()
sheet = wb.active

# Headers
sheet["A1"] = "date"
sheet["B1"] = "product"
sheet["C1"] = "Category"
sheet["D1"] = "Amount"
sheet["E1"] = "Total"

# Write data
for i in range(len(df)):
    sheet[f"A{i+2}"] = df["date"].iloc[i]
    sheet[f"B{i+2}"] = df["product"].iloc[i]
    sheet[f"C{i+2}"] = df["Category"].iloc[i]
    sheet[f"D{i+2}"] = df["Amount"].iloc[i]
    sheet[f"E{i+2}"] = df["Total"].iloc[i]

wb.save("Expense_Report_Final.xlsx")
```
